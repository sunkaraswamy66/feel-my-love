```python
emails['text'] = emails['text'].apply(lambda x: x.replace('Subject:',''))
emails['text'] = emails['text'].apply(lambda x: x.replace('subject:',''))
emails['text'] = emails['text'].apply(lambda x: x.replace('Subject',''))
emails['text'] = emails['text'].apply(lambda x: x.replace('subject',''))
```

```python
def process_text(text):
    import re
    
    return re.sub(r'\d',"",text)
```

```python
process_text('This is so good 12345 and here it is 4949')
```

```python
emails['text'] = emails['text'].apply(lambda x: process_text(x))
```

```python
emails
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
matrix = TfidfVectorizer(lowercase=True, 
                         stop_words='english', 
                         min_df = 0.2,
                         ngram_range=(1,2)
                         ,max_df=0.95
                        )
X = matrix.fit_transform(emails['text']).toarray()
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
matrix = CountVectorizer(lowercase=True, 
                         stop_words='english', 
                         min_df = 0.2,
                         ngram_range=(1,2)
                         ,max_df=0.95
                        )
X = matrix.fit_transform(emails['text']).toarray()
```

```python
X[0:5]
```

```python
matrix.get_feature_names_out()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, 
                                                    emails['spam'], 
                                                    train_size = 0.8, 
                                                    random_state = 12345)
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(n_estimators=200)
rf.fit(X_train,y_train)
rf_predict=rf.predict(X_test)
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(np.mean(rf_predict==y_test))
print(accuracy_score(y_test,rf_predict))
print(classification_report(y_test,rf_predict))
print(confusion_matrix(y_test,rf_predict))
```

```python
from xgboost import XGBClassifier
xg=XGBClassifier()
xg.fit(X_train,y_train)
xg_predict=xg.predict(X_test)
print(np.mean(xg_predict==y_test))
```
